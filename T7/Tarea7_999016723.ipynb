```python
import numpy as np
import pandas as pd
import glob
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import train_test_split
```

# Obtención, limpieza y transformacion de datos

```python
archivos = glob.glob('/content/drive/MyDrive/Postgrado/Mineria de datos/Redes Neuronales/Tarea7/datos/*.xlsx')

data = pd.concat([pd.read_excel(f) for f in archivos], ignore_index=True)

data = data.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
data = data[(data['VIC_SEXO'] != 9) & (data['AGR_TRABAJA'] != 9)]
data = data.dropna(subset=["VIC_ESCOLARIDAD"])
```

# Redes Neuronales

## Red Neuronal 1

### Variables dependientes e independientes

```python
X1 = data[['VIC_SEXO','VIC_EDAD', 'VIC_ESCOLARIDAD']]
X1 = X1.fillna(0)
Y1 = data['AGR_TRABAJA']

X1_train, X1_test, Y1_train, Y1_test = train_test_split(X1, Y1, test_size=0.2, random_state=42)
```

### Diseño y entreno de la red neuronal

```python
model1 = Sequential()
model1.add(Dense(8, input_dim=3, activation='relu'))
model1.add(Dense(4, activation='relu'))
model1.add(Dense(1, activation='sigmoid'))

model1.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

model1.fit(X1_train, Y1_train, epochs=50, batch_size=64, validation_data=(X1_test, Y1_test))

loss, accuracy = model1.evaluate(X1_test, Y1_test)
print(f'Loss: {loss}, Accuracy: {accuracy}') #93.69%
```

### Hechos/predicciones

```python
# Hecho 1. Hombre, 25 años, universitario
print(model1.predict(np.array([[1, 25, 59]])))
```
